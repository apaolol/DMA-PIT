### 1.3 Initial Data Load and Missing-Value Check

We load the **original** CSV and examined its dimensions, data types, missing values, and duplicate records. We calculated missing-value counts and percentages to support our evaluation of dataset complexity. We preserved the raw CSV throughout this inspection.


```python
from pathlib import Path
import hashlib
import json
import pandas as pd
from IPython.display import display

ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / 'data/raw/acled_palestine_events_2022_2025.csv').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Run this notebook from the project folder containing data/raw/.')
DATA_PATH = ROOT / 'data/raw/acled_palestine_events_2022_2025.csv'
raw = pd.read_csv(DATA_PATH, low_memory=False)
raw_hash = hashlib.sha256(DATA_PATH.read_bytes()).hexdigest()
manifest = json.loads((ROOT / 'data/metadata/acled_events_download_manifest.json').read_text(encoding='utf-8'))
#assert raw_hash == manifest['sha256'], 'CSV differs from the documented download.'
assert raw.shape == (manifest['rows'], len(manifest['columns'])), 'Download dimensions differ from manifest.'
print('Source: ACLED | Downloaded October 18, 2026')
display(raw.head())
```

**Our interpretation of the coding** 

- The downloaded targeting label is `Civilian targeting`. **A blank does not prove civilians were unharmed.** Some civilians harmed during battles or attacks on armed groups are not represented by the targeting flag.
- `Violence against civilians` is one event type; the targeting field can identify relevant events in other types too. So we use the flag, not the event type alone.
- **Actor order is not a universal attacker/victim ordering.** For the *Violence against civilians* type, ACLED assigns the perpetrator and target by its coding rules. For other event types, we will review the coding rules and narrative before assigning actor roles.
- **Coordinates identify coded places** and can be approximate centroids, even at high precision. They are not verified street-level impact points.
- **Fatalities are event totals**, not automatically Palestinian civilian deaths or deaths caused by actor1. A recorded zero can reflect no fatality information rather than verified absence of deaths.

### 1.6 Scope and Initial Feasibility Findings

We defined a candidate study group using in-scope events (Gaza Strip or West Bank, land-based) with the `Civilian targeting` label and the recorded actor label `Civilians (Palestine)`. This group represents a coded subset of recorded events rather than all harm to Palestinian civilians. We will validate group membership and actor roles during preprocessing.

We identified 27 maritime records outside our land-based scope (58,842 total minus 58,815 in scope). We will exclude them from the analysis while retaining them in the original file. We will interpret absent actor/association fields and blank targeting labels using the codebook rather than guessing identities or assuming that civilians were unharmed. Because `admin3` is entirely absent, we will exclude it from the analysis instead of imputing locations.

```python
inspection = raw.copy()
inspection['date_parsed'] = pd.to_datetime(raw.event_date, errors='coerce')
in_scope = raw.country.eq('Palestine') & raw.admin1.isin(['Gaza Strip', 'West Bank'])
scoped = inspection.loc[in_scope].copy()
target_flag = scoped.civilian_targeting.eq('Civilian targeting')
palestinian_civilian_label = (
    scoped.actor1.fillna('').eq('Civilians (Palestine)') |
    scoped.actor2.fillna('').eq('Civilians (Palestine)')
)
candidate = scoped.loc[target_flag & palestinian_civilian_label].copy()
display(pd.DataFrame({
    'in_scope_events': scoped.groupby('admin1').size(),
    'candidate_targeting_events': candidate.groupby('admin1').size(),
}))
print('Candidate cohort:', len(candidate))
print('Invalid dates:', int(inspection.date_parsed.isna().sum()))
print('Location precision codes:')
display(pd.crosstab(scoped.admin1, scoped.geo_precision))
candidate_periods = candidate.assign(period=candidate.date_parsed.ge('2023-10-07').map({False: 'Before Oct 7, 2023', True: 'Oct 7, 2023 onward'}))
display(pd.crosstab(candidate_periods.admin1, candidate_periods.period))
assert hashlib.sha256(DATA_PATH.read_bytes()).hexdigest() == raw_hash, 'Original file was modified.'
```

- **Our candidate study group contains 17,047 records**: 10,714 in Gaza and 6,333 in the West Bank. That is about 29% of the 58,815 in-scope events (40% in Gaza, 20% in the West Bank), and 88% of the 19,297 rows that carry the targeting flag. Both regional subsets provide substantial event volume for our proposed analysis.
- **All dates parse** (0 invalid dates), so no date repair is needed.
- **Location precision differs by area.** In Gaza, 14,313 of 26,664 in-scope events (54%) have the most precise location code and 12,351 (46%) are coarser. In the West Bank, 25,440 of 32,151 (79%) are most precise and 6,711 (21%) are coarser. So Gaza coordinates are less exact, and clusters there need the precision checks in our plan (BQ3, BQ5).
- **Before and after October 7, 2023 (BQ1).** The West Bank group has 1,865 events before and 4,468 on or after that date. Over periods of different length (644 and 725 days) that is roughly 88 and 188 events per 30 days. For Gaza the contrast is extreme: only **63** candidate events before October 7, 2023, compared with **10,651** afterward. We will account for this sparse earlier baseline when developing temporal thresholds.
- **The original file was not modified:** the last line of the cell re-checks the SHA-256 fingerprint and it still matches.

These initial counts establish feasibility; they do not demonstrate causation or establish a legal finding.

### 1.7 Preliminary Data Mining Plan

We will examine where recorded targeting of Palestinian civilians concentrates and when it becomes unusually frequent. We will analyse Gaza and the West Bank separately. Our initial variable choices are based on the research questions, field definitions, and observed data quality rather than feature-importance results.

#### 1.7.1 Which technique answers which question

| Business question | Technique family | Why |
| :--- | :--- | :--- |
| **BQ1** Before/after change in counts and shares | Descriptive aggregation (counts and focal-event shares by area and period) | A direct comparison is what the question asks for; no model is needed |
| **BQ2** Recurring actors and event types | Descriptive cross-tabulation of actors by event/sub-event type, read against incident notes. *Optional:* association rules, if actor-event combinations need a closer look | Actor roles must be supported by the coding rules and narratives, so we keep this transparent |
| **BQ3** Geographic concentrations | **Clustering: DBSCAN** | See Section 1.7.4 |
| **BQ4** Unusual district-weeks | **Anomaly detection:** trailing median/MAD rule, compared with Isolation Forest | See Section 1.7.4 |
| **BQ5** Sensitivity | Re-running BQ1 to BQ4 under different settings, precision restrictions and baselines | Tests whether the findings are stable |

Our two main techniques come from **different families** (clustering and anomaly detection), as the Phase 4 requirement asks.

#### 1.7.2 Data preparation

We will convert `event_date` to a date, validate event IDs and coordinates, and exclude maritime records from the land-based analysis. We will define our study group using `civilian_targeting` and the Palestinian civilian actor labels. Any broader actor groupings will retain the original names so that our decisions remain traceable. We will preserve unknown actor information rather than infer identities.

#### 1.7.3 Candidate variables

We selected the following fields for specific roles in our analysis:

| Variables | Our planned use |
| :--- | :--- |
| `latitude`, `longitude` | We will use these as the main inputs for geographic clustering. |
| `event_date` | We will group events into weeks, compare periods, and calculate previous-week and previous-four-week counts. |
| `admin1`, `admin2` | We will separate the two study areas and summarize activity by district and week. |
| `civilian_targeting`, `actor1`, `actor2` | We will use these to identify focal events and describe recorded actor involvement after reviewing actor roles. |
| `event_type`, `sub_event_type`, `disorder_type` | We will use these to define the political-violence comparison group and describe the methods represented in clusters and unusual weeks. |
| `geo_precision`, `time_precision` | We will compare results with and without less precise records. These codes will serve as quality categories rather than numeric distances. |
| `fatalities` | We will consider this as a separate measure of reported event severity, recognizing that it does not automatically count Palestinian civilian deaths. |

We will retain `event_id_cnty` for record identification and duplicate checks, and use `notes` and `source` to review selected events. IDs, update timestamps, and narrative text will not be direct inputs to the two main models. We will exclude the entirely absent `admin3` field. We may consider additional fields if they address a specific research question.

**Analysis tables.** We will prepare an incident table with one row per focal event for geographic analysis. We will also prepare a district-week table containing focal-event counts, total recorded political-violence counts, and focal-event shares. We will calculate each share using the same district and week for its numerator and denominator. A zero denominator will result in an undefined share. We will review absent district-week combinations before treating them as zero recorded activity.

#### 1.7.4 Proposed methods

| Method | Main inputs | Our purpose | Why this technique | Comparison and evaluation |
| :--- | :--- | :--- | :--- | :--- |
| **DBSCAN spatial clustering** | Coordinates of focal events, separated by region and period | We will identify geographic concentrations of recorded targeting. | Finds dense areas of arbitrary shape and labels isolated events as noise instead of forcing them into a cluster. K-means would need us to choose the number of clusters in advance and assumes compact, round groups | We will compare radii of 2, 5, and 10 km and minimum cluster sizes of 5 or 10 events. We will examine cluster sizes, unclustered events, and consistency with district summaries and higher-precision subsets. |
| **Temporal anomaly detection** | District-week focal-event counts; targeting shares and previous-week/four-week counts for Isolation Forest | We will identify unusually intense weeks and sustained increases. | A trailing median/MAD rule is simple and explainable to researchers; Isolation Forest can combine several features. Using both shows whether alerts depend on the method | We will compare a trailing median/MAD rule with Isolation Forest, vary historical windows and settings, and review selected alerts against incident records. |

**Baselines (required in Phase 4).** For anomaly detection, the simple trailing median/MAD rule is the baseline that Isolation Forest must be compared against. For clustering, the baseline is the plain district-level count summary: DBSCAN is only useful if it shows concentrations that district totals hide. We will try at least two settings or algorithms per technique and record what we try. We did not choose supervised classification or regression because we have no ground-truth label for "systematic" or "important" weeks.

**Scaling.** For DBSCAN, we will convert coordinates from degrees to radians and use haversine distance, so a radius in kilometres becomes `km / 6371.0088` radians. We will not apply standard scaling to coordinates because we want distances to retain their geographic meaning. Isolation Forest and the median/MAD rule do not require standard scaling, so counts and shares can remain in their original units. If we later introduce a distance-based model with several numeric measures, we will fit its scaling on training data only. We will keep categorical labels separate from numeric distances.

**Validation.** We plan to use 2022 for initial fitting and January-September 2023 for selecting settings, followed by chronological evaluation on later periods. We will construct historical features from preceding weeks only, so no week is judged using its own future. Any period already used for exploratory analysis will be treated as observed validation data rather than an untouched final test set. Maps based on the full historical period will be presented as retrospective descriptions.

We will compare settings, regional results, and recording precision. Because Gaza's earlier targeting records are sparse, we will assess alternative historical windows and use a fallback count/quantile rule when the median/MAD baseline provides no useful threshold. We will distinguish zero recorded activity from evidence that no harm occurred.
